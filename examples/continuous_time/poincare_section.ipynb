```python
import numpy as np
import matplotlib.pyplot as plt
from pynamicalsys import ContinuousDynamicalSystem as cds, PlotStyler
```

```python
cds.available_models()
```

```python
ds = cds(model="duffing")
```

```python
ds.info["equation"]
```

```python
delta = 0.2
alpha = 1
beta = 1
gamma = 0.5
omega = np.sqrt(2 * alpha)
parameters = [delta, alpha, beta, gamma, omega]

T = 2 * np.pi / omega
t_over_T = 40
sample_time = t_over_T * T
transient_time = 100 * T
total_time = sample_time + transient_time
num_samples = 100000
```

```python
u0 = [1, 0]
```

```python
trajectory = ds.trajectory(u0, total_time, parameters=parameters, transient_time=transient_time)
```

```python
strobe_map = ds.stroboscopic_map(u0, num_samples, T, parameters=parameters, transient_time=transient_time)
```

```python
ps = PlotStyler(fontsize=18, markersize=.25, markeredgewidth=0)
ps.apply_style()

fig, ax = plt.subplots(1, 2, sharex=True, sharey=True, figsize=(10, 4))

ax[0].plot(trajectory[:, 1], trajectory[:, 2], "-", c="k")
ax[1].plot(strobe_map[:, 1], strobe_map[:, 2], "o", c="k")
```
