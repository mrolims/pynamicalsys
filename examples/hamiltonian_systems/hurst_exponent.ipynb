```python
import numpy as np
import matplotlib.pyplot as plt
from pynamicalsys import PlotStyler, HamiltonianSystem as HS
```

```python
hs = HS(model="henon heiles")
```

```python
E = 1 / 8  # Total energy of the system
x = 0  # Define the initial condition
y = -0.16
py = 0
px = np.sqrt(2 * (E - x**2 * y + y**3 / 3) - x**2 - y**2 - py**2)
py_range = (-0.3, 0.3)
PY = np.linspace(*py_range, 200)

dof = 2
q = np.array([x, y])
p = np.array([px, py])

total_time = 2000
num_intersections = 5000
```

```python
PS_svy4 = hs.poincare_section(q, p, num_intersections)
hs.integrator("imp")
PS_imp = hs.poincare_section(q, p, num_intersections)
```

```python
plt.plot(PS_svy4[:, 2], PS_svy4[:, 4], "ko", markersize=1.2, markeredgewidth=0)
plt.plot(PS_imp[:, 2], PS_imp[:, 4], "ro", markersize=1.2, markeredgewidth=0)
```

```python
hs.integrator("imp")
hs.hurst_exponent(q, p, num_intersections)
```

```python

hs.integrator("svy4")
hs.hurst_exponent(q, p, num_intersections)
```

```python
MLE = np.zeros_like(PY)
for i in range(PY.shape[0]):
    py = PY[i]
    px = np.sqrt(2 * (E - x**2 * y + y**3 / 3) - x**2 - y**2 - py**2)
    q = [x, y]
    p = [px, py]
    MLE[i] = hs.lyapunov(q, p, total_time, num_exponents=1)
```

```python
H = np.zeros((PY.shape[0], 3))
for i in range(PY.shape[0]):
    py = PY[i]
    px = np.sqrt(2 * (E - x**2 * y + y**3 / 3) - x**2 - y**2 - py**2)
    q = [x, y]
    p = [px, py]
    H[i] = hs.hurst_exponent(q, p, num_intersections)
```

```python
ps = PlotStyler(fontsize=18)
ps.apply_style()

fig, ax = plt.subplots(2, 1, sharex=True, figsize=(10, 4))

ax[0].plot(PY, MLE, "k-")
ax[1].plot(PY, H[:, 0], "r-")
ax[1].plot(PY, H[:, 1], "g-")
ax[1].plot(PY, H[:, 2], "b-")
ax[1].plot(PY, H.mean(axis=1), "k-")
```
