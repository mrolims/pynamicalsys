```python
from pynamicalsys import DiscreteDynamicalSystem as dds
from pynamicalsys import PlotStyler
import numpy as np
import matplotlib.pyplot as plt
from numba import njit
```

```python
from pynamicalsys import DiscreteDynamicalSystem

ds = DiscreteDynamicalSystem(model="logistic map")
ds.set_parameters([3.9])

orbit, period = ds.build_pseudo_orbit(
    u=0.1,
    exits=[[0.0, 1.0]],
    total_time=10_000,
    refine="gradient",  # ou "sampling"
)

print(orbit.shape)
print(period)
```

```python
ps = PlotStyler()
ps.apply_style()

fig, ax = plt.subplots()
ax.plot(orbit[:-1],orbit[1:] , ',k')
ax.set_xlim(0,1)
plt.show()
```

```python
from pynamicalsys import DiscreteDynamicalSystem

ds = DiscreteDynamicalSystem(model="henon map")
ds.set_parameters([1.50, 0.3])

orbit, period = ds.build_pseudo_orbit(
    u=[0.1, 0.2],
    exits=[[-2, 2], [-2, 2]],
    total_time=50_000,
    refine="gradient",  # ou "sampling"
)

print(orbit.shape)
print(period)
```

```python
ps = PlotStyler()
ps.apply_style()

fig, ax = plt.subplots()
ax.plot(orbit[:, 0], orbit[:, 1], ',k')
plt.show()
```
